```python
# --- make csv file for spinup run inputs
import os

import pandas as pd
```

```python
spindict = {
    "site": ["site_311", "site_411"],
    "lat": [42.5, 32],
    "lon": [-91, -83],
    "mat": [8.22219, 18.52789],
    "soilmoisture": [0.282727, 0.231552],
    "qrun": [0.3513611, 0.2434256],
    "tsom": [2.051667, 2.276667],
    "erosion": [0.001013, 0.00084],
    "nitrif": [1.005952, 0.831883],
    "tph": [6.058007, 5.200242],
    "cec": [21.10329, 1.96125],
    "tec": [20.98031, 46.91557],
    "tsoilco2": [-1.80371, -1.61194],
    "poro": [0.447, 0.419],
    "alpha": [1.0, 1.0],
    "cec_adsorption_on": [True, True],
    "spinup_parentrock_file": ['parentlist9.json', 'parentlist9.json'],
    "zom": [0.15, 0.15],
    "initguess": ["DONOTUSE_s311_cecON_5m_spintuneup4_field", "DONOTUSE_s411_cecON_5m_spintuneup4_field"],    # dirname of spinup to use for guessing initial inputs
}
ztot_field = [0.25, 0.5, 1.0, 1.5, 2.0, 3.0, 4.0, 5.0]
```

## Programmatically add spinname and nz
nz = ztot_field * 60
spinname = site_{site}_pr9_cecON_{ztot_field}m

```python
# make base dataframe 
df_spin_base = pd.DataFrame(spindict)

# repeat the base dataframe for each ztot_field
df = pd.concat([df_spin_base.assign(ztot_field=val) for val in ztot_field], ignore_index=True)

# add 'nz'
df['nz'] = (df['ztot_field'] * 60).astype(int)

# add spinname
# f'{df['site']}_pr9_cecON_{df['ztot_field']}m'
df['spinname'] = df.apply(
    lambda row: f"{row['site']}_pr9_cecON_{row['ztot_field']}m",
    axis=1
)

df
```

```python
# --- save result
savehere = "s3://carbonplan-carbon-removal/ew-workflows-data/scepter/batch"
savefn = "spinup-inputs-allDepths-cecON.csv"
df.to_csv(os.path.join(savehere, savefn), index=False)
```
